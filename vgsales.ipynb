Countplot — How many games belong to each category?
What does it do?
A countplot counts how many times each category appears in a column.

For example, your dataset contains genres such as:
- Sports
- Action
- Racing
- Platform
- Role-Playing
You might want to know which genre has the most games.

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df,x='Genre')

plt.title('Number of Games by Genre')
plt.xlabel('Genre')
plt.ylabel('Number of Games')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

sns.countplot(data=df, x='Platform')

plt.xticks(rotation=90)
plt.title('Number of Games by Platform')
plt.show()
```

```python
plt.figure(figsize=(14, 5))

sns.countplot(data=df, x='Year')

plt.xticks(rotation=90)
plt.title('Number of Games Released Each Year')
plt.show()
```

Subplots — Display multiple graphs together
What are subplots?
Subplots allow you to display multiple graphs within a single figure.

Suppose you want to examine the distributions of:
- North American sales
- European sales
- Japanese sales
- Other regional sales
Instead of creating four separate figures, you can display them together.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(df['NA_Sales'], bins=30, kde=True, ax=axes[0, 0])
axes[0, 0].set_title('North American Sales Distribution')

sns.histplot(df['EU_Sales'], bins=30, kde=True, ax=axes[0, 1])
axes[0, 1].set_title('European Sales Distribution')

sns.histplot(df['JP_Sales'], bins=30, kde=True, ax=axes[1, 0])
axes[1, 0].set_title('Japanese Sales Distribution')

sns.histplot(df['Other_Sales'], bins=30, kde=True, ax=axes[1, 1])
axes[1, 1].set_title('Other Regional Sales Distribution')

plt.tight_layout()
plt.show()
```

Jointplot — Understand the relationship between two numeric variables
This is particularly useful when exploring numerical features.
What does it do?
A jointplot shows the relationship between two numeric variables.

It typically includes:
- A central scatterplot or another bivariate plot.
- A distribution of the first variable along the top.
- A distribution of the second variable along the side.
For your dataset, we can explore the relationship between North American and European sales.

```python
sns.jointplot(
    data=df,x='NA_Sales',y='EU_Sales',kind='scatter'
)
plt.show()
```

```python
sns.jointplot(
    data=df,
    x='NA_Sales',
    y='JP_Sales',
    kind='scatter',
    height=7
)

plt.show()
```

Heatmap — Identify correlations between numeric variables
A heatmap uses colors to represent values in a matrix.

In EDA, one common use is visualizing the correlation between numeric columns.

```python
numeric_df = df.select_dtypes(include='number')

corr_matrix = numeric_df.corr()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Video Game Sales')
plt.show()
```

```python
sales_columns = [
    'NA_Sales',
    'EU_Sales',
    'JP_Sales',
    'Other_Sales'
]

plt.figure(figsize=(8, 6))

sns.heatmap(
    df[sales_columns].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Regional Sales')
plt.show()
```

Bonus: Use a boxplot to detect outliers
Since you're learning EDA, I recommend learning boxplot alongside these graphs.

A boxplot helps you examine the distribution of numeric data and identify potential outliers.

```python
plt.figure(figsize=(10, 5))

sns.boxplot(data=df, x='Global_Sales')

plt.title('Distribution of Global Video Game Sales')
plt.show()
```

```python
yearly_sales = df.groupby('Year')['Global_Sales'].sum().reset_index()

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=yearly_sales,
    x='Year',
    y='Global_Sales'
)

plt.title('Total Global Video Game Sales by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Total Global Sales (Millions)')
plt.show()
```

| Graph | Best used for | Example from your dataset |
|---|---|---|
| `countplot()` | Counting categorical observations | Number of games per genre |
| `histplot()` | Examining a numeric distribution | Distribution of `Global_Sales` |
| `boxplot()` | Examining spread and potential outliers | Global sales distribution |
| `scatterplot()` | Exploring relationships between two numeric variables | `NA_Sales` vs `EU_Sales` |
| `jointplot()` | Exploring a numeric relationship and both marginal distributions | `NA_Sales` vs `JP_Sales` |
| `heatmap()` | Visualizing a matrix of numeric correlations | Correlations between sales regions |
| `subplots()` | Arranging multiple graphs in one figure | Comparing sales distributions across regions |
| `lineplot()` | Examining trends over an ordered variable | Total global sales by release year |
